## Check for outliers

### From min, max values we can there is negative values for brightness_std
### If it is standard deviation of brightness values, there can't be a negative value possible

```python
df[df["brightness_std"] < 0]
```

### These are wrong values. They need to be dropped

## Next check based on IQR fencing

```python
Q1 = df.quantile(0.25)
Q3 = df.quantile(0.75)
IQR = Q3 - Q1
lower_fence = Q1 - IQR
upper_fence = Q3 + IQR

outliers_mask1 = df < lower_fence
outliers_mask2 = df > upper_fence
```

```python
print(outliers_mask1.sum(axis = 0))
print("\n\n----------------------------------------------x----------------------------------------------\n\n")
print(outliers_mask1.sum(axis = 1).sum())
```

```python
print(outliers_mask2.sum(axis = 0))
print("\n\n----------------------------------------------x----------------------------------------------\n\n")
print(outliers_mask2.sum(axis = 1))
```

```python
df_temp = outliers_mask2.sum(axis = 1)
df_temp.max()
```

```python
print(df_temp[df_temp > 0].shape)
print(df_temp[df_temp > 1].shape)
print(df_temp[df_temp > 2].shape)
print(df_temp[df_temp == 1].shape)
print(df_temp[df_temp == 2].shape)
print(df_temp[df_temp == 3].shape)
```

```python
df_temp[(df_temp > 0) & (df["confidence_score"] < 0.7)].shape
```
